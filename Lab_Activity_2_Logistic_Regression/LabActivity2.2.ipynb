Q2. For the given dataset, investigate whether a person is susceptible to a certain disease or not.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt
import seaborn as sn
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
data=pd.read_csv('Logistic_Regression_Dataset_Exercise_2.csv')
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
data.shape
```

```python
data['Disease'] = data['Disease'].map({'diseased': 1, 'not diseased': 0})
data['Gender'] = data['Gender'].map({'Male': 1, 'Female': 0})
data['Smoker_status'] = data['Smoker_status'].map({'Smoker': 1, 'Non-smoker': 0})
```

```python
feature_cols=['Age', 'Gender', 'Smoker_status']
X=data[feature_cols]
Y=data['Disease']
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=5) 
```

```python
model = LogisticRegression()
model.fit(X_train,Y_train)
```

```python
model_predicted=model.predict(X_test)
```

```python
print("Actual vs Predicted:")
print(pd.DataFrame({'Actual':Y_test.values,'Predicted':model_predicted}))
```

```python
acc=accuracy_score(Y_test,model_predicted)
print("Classification Evaluation Metrics")
print(f"Accuracy: {acc:.4f}")
print("Confusion Matrix:")
print(confusion_matrix(Y_test,model_predicted))
print("Classification Report:")
print(classification_report(Y_test,model_predicted))
```

```python
confusion_matrix=pd.crosstab(Y_test,model_predicted,rownames=['Actual'],colnames=['Predicted'])
sn.heatmap(confusion_matrix, annot=True)
```
