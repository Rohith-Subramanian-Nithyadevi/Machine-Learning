**Q2.** Calculate the distance between a new entry and other existing values using the Euclidean distance formula.

| Brightness | Saturation | Class |
|---|---|---|
| 40 | 20 | Red |
| 50 | 50 | Blue |
| 60 | 90 | Blue |
| 10 | 25 | Red |
| 70 | 70 | Blue |
| 60 | 10 | Red |
| 25 | 80 | Blue |

**Dataset:** The small table given in the exercise (built directly in this notebook), following the same programming approach used in Section 9.5 of the Lab Activity Book (Implementing KNN using Python).

### Importing Required Libraries

```python
import numpy as np
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
```

### Reading the Dataset (given in the exercise)

```python
data = pd.DataFrame({
    'Brightness': [40, 50, 60, 10, 70, 60, 25],
    'Saturation': [20, 50, 90, 25, 70, 10, 80],
    'Class':      ['Red', 'Blue', 'Blue', 'Red', 'Blue', 'Red', 'Blue']
})
print(data.shape)
data
```

### Checking for any null values in dataset

```python
data.isnull().sum()
```

### Assigning Dependent and Independent Variables

```python
x = data[['Brightness', 'Saturation']]
y = data['Class']
```

### New Entry to Classify

```python
new_point = np.array([45, 45])
print('New entry (Brightness, Saturation): ', new_point)
```

### Calculating the Euclidean Distance from the New Entry to every Existing Point

```python
distances = np.sqrt(((x.values - new_point) ** 2).sum(axis=1))
dist_table = data.copy()
dist_table['Euclidean_Distance_to_New_Entry'] = distances
dist_table.sort_values(by='Euclidean_Distance_to_New_Entry')
```

### Training the Model using KNN (Euclidean / minkowski, p=2)

```python
model = KNeighborsClassifier(n_neighbors=3, metric='minkowski', p=2)
model.fit(x, y)
predicted_class = model.predict([new_point])
print('Predicted class for the new entry', tuple(new_point), ':', predicted_class[0])
```
